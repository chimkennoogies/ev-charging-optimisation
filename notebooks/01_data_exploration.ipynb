# Data Exploration

This notebook explores the EV charging dataset to understand:
- Data structure and quality
- Feature distributions
- Relationships between variables
- Missing values and outliers
- Target variable characteristics

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
from pathlib import Path

# Add parent directory to path
sys.path.append(str(Path().resolve().parent))

from src.utils import load_data

# Set style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
%matplotlib inline
```

```python
# Load data
df = load_data("ev_charging_data.csv")
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# Basic information
df.info()
```

```python
# Statistical summary
df.describe()
```

```python
# Check for missing values
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Percentage': missing_pct
})
missing_df[missing_df['Missing Count'] > 0].sort_values('Missing Count', ascending=False)
```

```python
# Target variable distribution
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

axes[0].hist(df['Charging Cost (USD)'], bins=50, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Charging Cost (USD)')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Distribution of Charging Cost')

axes[1].boxplot(df['Charging Cost (USD)'], vert=True)
axes[1].set_ylabel('Charging Cost (USD)')
axes[1].set_title('Box Plot of Charging Cost')

plt.tight_layout()
plt.show()
```

```python
# Categorical features
categorical_cols = df.select_dtypes(include=['object']).columns
print("Categorical features:")
for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].value_counts().head(10))
```

```python
# Correlation matrix for numerical features
numerical_cols = df.select_dtypes(include=[np.number]).columns
correlation_matrix = df[numerical_cols].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Matrix of Numerical Features')
plt.tight_layout()
plt.show()
```

```python
# Relationship between key features and target
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

axes[0, 0].scatter(df['Energy Consumed (kWh)'], df['Charging Cost (USD)'], alpha=0.5)
axes[0, 0].set_xlabel('Energy Consumed (kWh)')
axes[0, 0].set_ylabel('Charging Cost (USD)')
axes[0, 0].set_title('Energy vs Cost')

axes[0, 1].scatter(df['Charging Duration (hours)'], df['Charging Cost (USD)'], alpha=0.5)
axes[0, 1].set_xlabel('Charging Duration (hours)')
axes[0, 1].set_ylabel('Charging Cost (USD)')
axes[0, 1].set_title('Duration vs Cost')

axes[1, 0].scatter(df['Charging Rate (kW)'], df['Charging Cost (USD)'], alpha=0.5)
axes[1, 0].set_xlabel('Charging Rate (kW)')
axes[1, 0].set_ylabel('Charging Cost (USD)')
axes[1, 0].set_title('Rate vs Cost')

axes[1, 1].scatter(df['Battery Capacity (kWh)'], df['Charging Cost (USD)'], alpha=0.5)
axes[1, 1].set_xlabel('Battery Capacity (kWh)')
axes[1, 1].set_ylabel('Charging Cost (USD)')
axes[1, 1].set_title('Battery Capacity vs Cost')

plt.tight_layout()
plt.show()
```

```python
# Cost by charger type
if 'Charger Type' in df.columns:
    plt.figure(figsize=(10, 6))
    df.boxplot(column='Charging Cost (USD)', by='Charger Type', ax=plt.gca())
    plt.title('Charging Cost by Charger Type')
    plt.suptitle('')  # Remove default title
    plt.xlabel('Charger Type')
    plt.ylabel('Charging Cost (USD)')
    plt.tight_layout()
    plt.show()
```

```python
# Cost by time of day
if 'Time of Day' in df.columns:
    plt.figure(figsize=(12, 6))
    df.boxplot(column='Charging Cost (USD)', by='Time of Day', ax=plt.gca())
    plt.title('Charging Cost by Time of Day')
    plt.suptitle('')
    plt.xlabel('Time of Day')
    plt.ylabel('Charging Cost (USD)')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```
